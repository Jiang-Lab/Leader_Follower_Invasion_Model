## Plot the data using 3D plotting

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
file_path = "/Users/sheriffakeeb/Desktop/Sample_Simulation/Outputs/infiltrative_Area.csv"  
df = pd.read_csv(file_path)

# Compute the average invasive area for each unique combination of parameters
df_avg = df.groupby(["Adhesion Energy", "Chemotaxis Lambda", "Proliferative Probability"], as_index=False)["Infiltrative Area"].mean()

# Extract relevant columns
adhesion_energy = df_avg["Adhesion Energy"].values
chemotaxis_lambda = df_avg["Chemotaxis Lambda"].values
proliferative_probability = df_avg["Proliferative Probability"].values
invasive_area = df_avg["Infiltrative Area"].values

# Define grid size for interpolation
Nx, Ny, Nz = 10, 10, 10  # Adjusting for computational efficiency
X, Y, Z = np.meshgrid(np.linspace(adhesion_energy.min(), adhesion_energy.max(), Nx),
                       np.linspace(chemotaxis_lambda.min(), chemotaxis_lambda.max(), Ny),
                       -np.linspace(proliferative_probability.min(), proliferative_probability.max(), Nz))

data = invasive_area.reshape((Nx, Ny, Nz), order='F')

kw = {
    'vmin': data.min(),
    'vmax': data.max(),
    'levels': np.linspace(data.min(), data.max(), 10),
}

# Create a figure with 3D ax
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Plot contour surfaces
_ = ax.contourf(
    X[:, :, 0], Y[:, :, 0], data[:, :, 0],
    zdir='z', offset=0, **kw
)
_ = ax.contourf(
    X[0, :, :], data[0, :, :], Z[0, :, :],
    zdir='y', offset=chemotaxis_lambda.min(), **kw
)
C = ax.contourf(
    data[:, -1, :], Y[:, -1, :], Z[:, -1, :],
    zdir='x', offset=adhesion_energy.max(), **kw
)

# Set limits of the plot from coord limits
ax.set(
    xlim=[X.min(), X.max()],
    ylim=[Y.min(), Y.max()],
    zlim=[Z.min(), Z.max()],
    xlabel='Contact Energy',
    ylabel='Chemotaxis Lambda',
    zlabel='Proliferative Probability',
)

# Set view angle
ax.view_init(40, -30)

# Colorbar
fig.colorbar(C, ax=ax, fraction=0.02, pad=0.1, label='Infiltrative Area')

# Show 3D Figure
plt.show()

# --- 2D Contour Plots ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Contact Energy vs Chemotaxis Lambda
cont1 = axes[0].tricontourf(adhesion_energy, chemotaxis_lambda, invasive_area, levels=10, cmap='viridis')
axes[0].set_xlabel('Contact Energy')
axes[0].set_ylabel('Chemotaxis Lambda')
axes[0].set_title('Contact Energy vs Chemotaxis Lambda')
fig.colorbar(cont1, ax=axes[0], label='Infiltrative Area')

# Chemotaxis Lambda vs Proliferative Probability
cont2 = axes[1].tricontourf(chemotaxis_lambda, proliferative_probability, invasive_area, levels=10, cmap='viridis')
axes[1].set_xlabel('Chemotaxis Lambda')
axes[1].set_ylabel('Proliferative Probability')
axes[1].set_title('Chemotaxis Lambda vs Proliferative Probability')
fig.colorbar(cont2, ax=axes[1], label='Infiltrative Area')

# Contact Energy vs Proliferative Probability
cont3 = axes[2].tricontourf(adhesion_energy, proliferative_probability, invasive_area, levels=10, cmap='viridis')
axes[2].set_xlabel('Contact Energy')
axes[2].set_ylabel('Proliferative Probability')
axes[2].set_title('Contact Energy vs Proliferative Probability')
fig.colorbar(cont3, ax=axes[2], label='Infiltrative Area')


# Show plots
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.interpolate import griddata
from matplotlib.colors import ListedColormap

# Load the dataset
file_path = "/Users/sheriffakeeb/Desktop/Sample_Simulation/Outputs/contact_vs_chemotaxis.csv" 
df = pd.read_csv(file_path)

# Extract unique phenotypes in the correct order
phenotype_order = ["No invasion", "Single cell invasion", "Collective invasion, no break-off", "Single and cluster break off"]

# Assign specific colors to phenotypes in the correct order
phenotype_colors = {
    "No invasion": "green",
    "Single cell invasion": "blue",
    "Collective invasion, no break-off": "pink",
    "Single and cluster break off": "red"
}

# Create a colormap with the defined colors in the correct order
ordered_colors = [phenotype_colors[phenotype] for phenotype in phenotype_order]
custom_cmap = ListedColormap(ordered_colors)

# Convert phenotype labels into numerical categories
df["Phenotype"] = pd.Categorical(df["Phenotype"], categories=phenotype_order, ordered=True)
df["Phenotype Category"] = df["Phenotype"].cat.codes

# Create a grid for contour plotting
x_grid = np.linspace(df["Contact Energy"].min(), df["Contact Energy"].max(), 100)
y_grid = np.linspace(df["Chemotaxis Lambda"].min(), df["Chemotaxis Lambda"].max(), 100)
X_grid, Y_grid = np.meshgrid(x_grid, y_grid)

# Interpolate phenotype regions onto the grid using 'nearest' method
Z_grid = griddata((df["Contact Energy"], df["Chemotaxis Lambda"]), df["Phenotype Category"],
                  (X_grid, Y_grid), method='nearest')

# Create the contour plot
plt.figure(figsize=(8, 6))
contour = plt.contourf(X_grid, Y_grid, Z_grid, levels=np.arange(-0.5, len(phenotype_order), 1), cmap=custom_cmap, alpha=0.6)

# Add contour lines for better visualization
plt.contour(X_grid, Y_grid, Z_grid, levels=np.arange(-0.5, len(phenotype_order), 1), colors='black', linewidths=0.5)

# Label axes and title
plt.xlabel("Contact Energy")
plt.ylabel("Chemotaxis Lambda")
plt.title("Phase Plot: Contact Energy Vs Chemotas Lambda")

# Create a colorbar with phenotype labels
cbar = plt.colorbar(contour, ticks=range(len(phenotype_order)))
cbar.ax.set_yticklabels(phenotype_order)

# Show grid for better readability
plt.grid(True)

# Display the plot
plt.show()
```
